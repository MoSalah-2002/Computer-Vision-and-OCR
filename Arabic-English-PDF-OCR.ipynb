```python
import fitz  # PyMuPDF
from pdf2image import convert_from_path
import easyocr
import tempfile
import os
import numpy as np
```

```python
reader = easyocr.Reader(['ar', 'en'],gpu=True)
```

```python
input_pdf = "/content/x.pdf"
output_txt = "final_output.txt"

confidence_threshold = 0.30


pages = convert_from_path(input_pdf, dpi=300)
```

```python
# File to store all extracted text
with open(output_txt, "w", encoding="utf-8") as outfile:

    for page_num, img in enumerate(pages, start=1):
        print(f"OCR (GPU) processing page {page_num}...")

        # OCR on GPU
        results = reader.readtext(np.array(img))

        # Add page title
        outfile.write(f"\n\n--- Page {page_num} ---\n\n")

        # Extract text per page with confidence
        for (bbox, text, confidence) in results:
            if confidence >= confidence_threshold:
                outfile.write(text + "\n")
            else:
                print(f"Skipped low-confidence text: '{text}' ({confidence:.2f})")

print(f"Done! All text saved in: {output_txt}")
```
