```python
def get_numbers_from_img(id):
  ref = cv2.threshold(id, 0, 255, cv2.THRESH_BINARY_INV|cv2.THRESH_OTSU)[1]
  # plt.imshow(ref,cmap="gray")
  # plt.show()
  kernal = np.ones((3,3), np.uint8)
  ref = cv2.morphologyEx(ref, cv2.MORPH_OPEN, kernal)
  refCnts = cv2.findContours(ref.copy(), cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)
  refCnts = imutils.grab_contours(refCnts)
  refCnts = contours.sort_contours(refCnts, method="left-to-right")[0]
  # print(len(refCnts))
  digits = {}

  for (i, c) in enumerate(refCnts):
      (x, y, w, h) = cv2.boundingRect(c)
      roi = id[y-6:y + h+6, x-6:x + w+6]
      # cv2.rectangle(id,(x-6,y-6),(x + w+6,y + h+6),(0,255,0),2)
      roi = cv2.resize(roi, (57, 88))
      roi=cv2.bilateralFilter(roi,11,17,17)
      digits[i] = roi
      # plt.imshow(roi,cmap="gray")
      # plt.axis('off')
      # plt.show()

  return digits

img2=cv2.imread("/content/output_image (1).jpg")
img2 = cv2.resize(img2, (700, 480))
img2=img2[:,:685]
gray2=cv2.cvtColor(img2,cv2.COLOR_BGR2GRAY)
id = gray2[360:430 , 300:]
# plt.imshow(id)
# plt.show()
get_numbers_from_img(id)
```
